```python
import pandas as pd

# 1. Membaca file asli
# Koreksi: Menggunakan nama file yang benar 'Hasil_Sentimen_Final.xlsx' dan fungsi pd.read_excel
file_input = 'Hasil_Sentimen_Final.xlsx'
df = pd.read_excel(file_input)

# Memberi nama kolom secara eksplisit setelah membaca Excel
# Ini menggantikan behavior `names` dan `header` dari pd.read_csv sebelumnya
df.columns = ['kalimat', 'sentimen']

# 2. Fungsi Pembersihan Label (Menghilangkan tulisan "LABEL")
def clean_label(label):
    label = str(label).upper()
    if 'LABEL_4' in label or 'POSITIF' in label:
        return 'positif'
    elif 'LABEL_3' in label or 'NETRAL' in label:
        return 'netral'
    else:
        return 'negatif'

df['sentimen'] = df['sentimen'].apply(clean_label)

# 3. Menambah data menjadi 400 kalimat (Upsampling)
if len(df) < 400:
    multiplier = (400 // len(df)) + 1
    df_resampled = pd.concat([df] * multiplier, ignore_index=True)
    df_final = df_resampled.iloc[:400].copy()
else:
    df_final = df.iloc[:400].copy()

# 4. Mengacak urutan (Shuffle) agar tidak berurutan
df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)

# 5. SIMPAN KE EXCEL (.xlsx) - Ini cara paling ampuh agar tidak nyatu
df_final.to_excel('Hasil_Sentimen_Final_400.xlsx', index=False)

# Simpan juga ke CSV dengan titik koma (Format Excel Indonesia)
df_final.to_csv('Hasil_Sentimen_Final_400.csv', index=False, sep=';')

print("Selesai! File 'Hasil_Sentimen_Final_full.xlsx' telah dibuat.")
print("Sekarang kolom 'kalimat' dan 'sentimen' sudah terpisah dengan sempurna.")
```
